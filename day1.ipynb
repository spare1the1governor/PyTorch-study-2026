```python
import torch.nn.functional as F
x = torch.tensor([[1.0,2.0],[1.0,2.0],[1.0,2.0]])
y = torch.tensor([[1.0,0.0,1.0],[0.0,1.0,0.0],[0.0,1.0,1.0]])
w = torch.randn(2,3 , requires_grad = True)
b = torch.randn(1,3 , requires_grad = True)
z = x @ w + b
p = torch.sigmoid(z) # задание ф-ии активации
loss = F.binary_cross_entropy(p, y) # задание ф-ии ошибки 

loss.backward()
print("prediction size z:", z.shape)
print("size y:", y.shape)
print("Loss:", loss.item())
```

```python
print(f"Gradient function for z = {z.grad_fn}")
print(f"Gradient function for loss = {loss.grad_fn}")
```

```python
print(z.requires_grad)
print(z.detach().requires_grad)
```
