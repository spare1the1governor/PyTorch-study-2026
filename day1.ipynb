```python
small_rand_tensor = torch.rand_like(small_tensor, dtype=torch.float)
small_one_tensor = torch.ones_like(small_tensor)
small_zero_tensor = torch.zeros_like(small_tensor)

print(f'{small_rand_tensor},\n {small_one_tensor},\n {small_zero_tensor}')
```

```python
rand_tensor = torch.rand(2,5)
ones_tensor = torch.ones(2,3)
zeros_tensor = torch.zeros(3,5)
full_tensor = torch.full((4, 4), 61)
print(f'{rand_tensor},\n {ones_tensor},\n {zeros_tensor},\n {full_tensor}')
```

### Tensors slices like in nupy

```python
big_tensor = [[[[1,1,4],[4,5,6]],[[2,3,4],[8,8,9]]],
              [[[3,3,4],[1,5,1]],[[2,1,1],[0,0,9]]]]
big_tensor = torch.tensor(big_tensor)
print(big_tensor.shape)
baby_tensor1 = big_tensor[0,:,1,:]
baby_tansor2 = big_tensor[1,1,:,2]
print(f'{baby_tensor1}, \n{baby_tansor2}')
```

```python
print(baby_tensor1.dtype)
baby_tensor1 @ baby_tensor1.T
big_tensor.reshape(6,4)
```

### Checkind and choosing device 

```python
if torch.cuda.is_available():
    print("CUDA (GPU) avellible! Shifting tensor opun GPU.")
    device = torch.device("cuda")
    tensor = baby_tensor1.to(device)
    print(f"Now tensor on : {tensor.device}")
    
else:
    print("GPU not avellible . Tensor stay opun CPU.")
```

###  Operations that store the result into the operand are called in-place.
